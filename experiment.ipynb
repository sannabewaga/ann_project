```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
import pickle
```

```python
data = pd.read_csv('Churn_Modelling.csv')
```

```python
data.head()
```

## preprocess the data
### drop irrelevant features

```python
data.drop(columns=['RowNumber','CustomerId','Surname'],axis= 1 , inplace=True)
```

```python
data.head()
```

```python
## encoding
```

```python
label_encoder_gender = LabelEncoder()
data['Gender']= label_encoder_gender.fit_transform(data['Gender'])
```

```python
data.head()
```

```python
##one hot encoding
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder()
geo_encoded = encoder.fit_transform(data[['Geography']])
geo_encoded
```

```python
encoder.get_feature_names_out(['Geography'])
```

```python
geo_encoded_df = pd.DataFrame(geo_encoded.toarray(),columns=encoder.get_feature_names_out(['Geography']))
```

```python
geo_encoded_df
```

```python
## combine data

data = pd.concat([data.drop(columns='Geography',axis=1),geo_encoded_df],axis=1)
data.head()
```

```python
with open('label_encoder_gender.pkl','wb') as file:
    pickle.dump(label_encoder_gender,file)

with open('ohe_geo.pkl','wb') as file:
    pickle.dump(encoder,file)

```
